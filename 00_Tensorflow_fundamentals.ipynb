```python
# Creating random tensors with tensorflow
random_1 = tf.random.Generator.from_seed(8)
random_1 = random_1.normal(shape =(3,2))
random_2 = tf.random.Generator.from_seed(8)
random_2 = random_2.normal(shape=(3,2))

# Are they equal
random_1, random_2, random_1 == random_2
```

```python
# Try shuffling the elements of a tensor
shuffled_tensor = tf.random.shuffle(random_1)
shuffled_tensor
```

Why would you want to shuffle the elements of a tensor?


Shuffling elements in a tensor is a crucial data preparation step in machine learning, primarily designed to improve model training performance and generalization. It breaks the ordering of data, ensuring that the model does not learn unwanted patterns related to the sequence of inputs.

```python
# Not shuffled
not_shuffled = tf.constant([[10,7],
                            [3,4],
                            [2,5]])

# Shuffle our non_shuffled tensor
tf.random.set_seed(42)
shuffled_tensor = tf.random.shuffle(not_shuffled, seed=42)

# compare the tensors
not_shuffled , shuffled_tensor
```

### Manipulating tensors

**Basic Operation**

`+`, `-`, `*`, `/`

```python
# You can add values to a tensor using the addition operator
tensor = tf.constant([[10, 7],[3,4]])
tensor+10
```

```python
# Multiplication also works
tensor *10
```

```python
# Create a tensor (3,2)
x = tf.constant([[1,2],
                 [3,4],
                 [5,6]])

# Create another tensor (3,2)
y = tf.constant([[7,8],
                 [9,10],
                 [11,12]])

# Matrix multiply tensors of the same shape
x @ y
```

```python
# Try using matmul
tf.matmul(x,y)
```

### Rules for matrix multiplication

*   The inner dimension must match
*   The resulting matrix must have the shape of the inner dimension
